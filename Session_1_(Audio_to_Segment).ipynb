```
# @title 2. Transcribe Audio to Segments (with Language Selection)
# This script loads an audio file, transcribes it using Whisper, and saves the text segments to a file for Session 2.

# 1. Install necessary libraries
# We use the latest version of whisper from github

# 2. Imports
import whisper
import librosa
import pickle
from google.colab import files

# 3. Upload Audio File
print("▶ Upload an audio file (mp3, wav, etc.):")
aud_up = files.upload()
if not aud_up:
    raise RuntimeError("Operation cancelled. No audio file was uploaded.")
audio_path = next(iter(aud_up))
print(f"✅ Audio file received: {audio_path}")

# 4. Prompt for Language Code (RESTORED)
# Prompt the user for the language code. If left blank, it defaults to None for auto-detection.
language_prompt = (
    "\nEnter the two-letter language code for the audio "
    "('hi' for Hindi, 'pa' for Punjabi, 'en' for English, etc.), "
    "or leave blank to auto-detect: "
)
language_code = input(language_prompt).strip() or None

# 5. Transcribe and Process Audio
print("\n🔊 Loading Whisper model and transcribing audio...")
# Using the "medium" model as in the original script.
# For faster but less accurate results, you could use "base" or "small".
model = whisper.load_model("medium")

# Prepare arguments for the transcription function
# This cleanly adds the 'language' argument only if the user provided one.
transcribe_args = {}
if language_code:
    transcribe_args['language'] = language_code
    print(f"   Transcribing in language: {language_code}")
else:
    print("   No language specified. Using auto-detection.")

# Perform the transcription
res = model.transcribe(audio_path, **transcribe_args)
segments = res["segments"]

# Get the total duration of the audio using librosa
duration = librosa.get_duration(path=audio_path)
print("✅ Transcription complete.")

# 6. Save Data for Session 2
# We save the segments and duration into a single file using pickle.
output_data = {
    'segments': segments,
    'duration': duration
}
output_filename = 'transcription_data.pkl'
with open(output_filename, 'wb') as f:
    pickle.dump(output_data, f)

print(f"\n✅ Transcription data successfully saved to '{output_filename}'.")
print("You can now download this file and upload it in Session 2.")
```
